## CARI DATA 100 SAMPEL

```python
import numpy as np
import pandas as pd

# Menetapkan seed agar angka acak yang dihasilkan selalu sama setiap kali kode dijalankan
np.random.seed(42)

# Membuat 100 data sampel acak (rentang angka yang dipakai: 1-100)
data = np.random.randint(1, 100, size=100)

# Menampilkan data sampel ke layar
print("Data sampel (100 data):")
print(data)
```

## ESTIMASI TITIK DAN INTERVAL (TINGKAT KESALAHAN = 5%)

```python
#ESTIMASI TITIK (Cari Mean dari Data Sampel)

point_estimate = np.mean(data)
print("Estimasi Titik (Rata-rata):", point_estimate)

#Menghitung jumlah total data sampel (n = 100)
n = len(data)

#Menghitung standar deviasi sampel (ddof=1 menggunakan rumus sampel n-1)
std_dev = np.std(data, ddof=1)

#Menentukan nilai Z untuk tingkat kepercayaan 95% (karena tingkat kesalahan/alpha = 5%)
z_score = 1.96

#Menghitung batas toleransi error (Margin of Error / MoE)
moe = z_score * (std_dev / np.sqrt(n))

#ESTIMASI INTERVAL
#Menghitung batas bawah dari interval kepercayaan
lower_bound = point_estimate - moe

#Menghitung batas atas dari interval kepercayaan
upper_bound = point_estimate + moe

#Menampilkan hasil interval kepercayaan 95%
print(f"Interval Kepercayaan 95%: [{lower_bound:.4f}, {upper_bound:.4f}]")
```

## RESAMPLING (JACKKNIFE & BOOTSTRAPPING)

```python
#METODE JACKKNIFE (Membuang 1 data secara bergantian sebanyak jumlah data)
jackknife_means = [] #Menyiapkan wadah kosong untuk menampung rata-rata jackknife

#Melakukan perulangan (loop) sebanyak total data (100 kali)
for i in range(len(data)):
    #Menghapus satu elemen data pada indeks ke-i secara bergantian
    resample_data = np.delete(data, i)

    #Menghitung rata-rata dari sisa 99 data, lalu memasukkannya ke dalam list
    jackknife_means.append(np.mean(resample_data))

#Menghitung rata-rata total dari seluruh hasil iterasi Jackknife
print(f"Rata-rata hasil Jackknife: {np.mean(jackknife_means):.4f}")

#METODE BOOTSTRAPPING (Mengambil sampel acak dengan pengembalian (replace=True))
n_iterations = 1000  #Menentukan jumlah iterasi sebanyak 1000 kali sesuai instruksi
bootstrap_means = [] #Menyiapkan wadah kosong untuk menampung rata-rata bootstrap

#Melakukan perulangan sebanyak 1000 kali iterasi
for _ in range(n_iterations):
    # Mengambil sampel acak sebanyak 100 data dengan pengembalian (data boleh kembar/terpilih ulang)
    resample_boot = np.random.choice(data, size=len(data), replace=True)

    #Menghitung rata-rata dari sampel bootstrap tersebut, lalu menyimpannya ke list
    bootstrap_means.append(np.mean(resample_boot))

#Menghitung rata-rata total dari 1000 hasil putaran bootstrapping
print(f"Rata-rata hasil Bootstrapping (1000 iterasi): {np.mean(bootstrap_means):.4f}")
```
